# redshift_notebooks quickstart — SQL cells against Redshift

Prerequisites:
- Dataeng-Prod AWS Client VPN connected (needed by the SSO connection)
- Kernel set to this package's `.venv` (kernel picker, top-right) — install it
  first with `uv venv` + `uv pip install -e .` from `src/redshift_notebooks/`
- Your project's dbt connection package installed in that same environment, and a
  `~/.config/redshift_notebooks/config.toml` pointing at it (only required if
  you use its connection factory below; the core library itself doesn't
  need it — see "Bring your own connection" in the README)

`register()` loads the jupysql `%sql`/`%%sql` magics and makes the engine the
active connection for the rest of the notebook, so every `%%sql` cell below
queries Redshift directly and renders its result as a table — no more
`conn.execute(query).fetch_dataframe()` per cell.

```python
import redshift_notebooks
from redshift_notebooks.adapters.config import get_configured_engine

engine = get_configured_engine()
redshift_notebooks.register(engine)
```

The cell above opens a browser window for Azure AD login on first use (same
tenant as DBeaver). Once registered, write plain SQL in `%%sql` cells:

```python
%%sql
select current_user, current_database()
```
