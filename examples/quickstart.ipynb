# sql-notebook-kit quickstart

Install `sql-notebook-kit[duckdb,viz]==0.1.0` in this notebook's kernel. This quickstart uses an isolated in-memory DuckDB database and needs no credentials or external data.

```python
from sql_notebook_kit import create_session

session = create_session(
    backend="duckdb",
    connection_kwargs={"database": ":memory:"},
)
session.register(visualization=True, max_rows=100)
```

```python
%%sql
select *
from (values (1, 'alpha'), (2, 'beta'), (3, 'gamma'))
as quickstart(amount, category)
order by amount
```

The SQL cell returns a bounded `NotebookResult` and exposes the eligible source query as `_df`. Use `_.dataframe` for the eager rows, `_.visualize()` for the workspace, or `_df.collect()` to rerun the source through the separately owned lazy connection.
